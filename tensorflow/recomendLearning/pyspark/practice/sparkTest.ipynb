```python
# Stop existing session
SparkSession.builder.getOrCreate().stop()
```

```python
# Initialize SparkSession
spark = SparkSession.builder \
    .appName("ExampleApp") \
    .master("local[*]") \
    .getOrCreate()

# Example DataFrame operation
data = [("John", 30), ("Doe", 25)]
df = spark.createDataFrame(data, ["Name", "Age"])

df.show()
```
